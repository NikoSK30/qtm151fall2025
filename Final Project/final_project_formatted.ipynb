##### Here, we group the data by year, the column we created in last step, and by their nationality. Since we merged all the year datatsets into one, it made this step more efificient by only sorting through necessary columns in one dataset rather than many. 

```python
yearly_avg = df_all.groupby(["Year", "nationality_name"])[["shooting","passing","defending"]].mean()
print(yearly_avg)

top5_shooting = (yearly_avg.groupby("nationality_name")["shooting"].mean().sort_values(ascending=False).head(5).index.tolist())

top5_passing = (yearly_avg.groupby("nationality_name")["passing"].mean().sort_values(ascending=False).head(5).index.tolist())

top5_defending = (yearly_avg.groupby("nationality_name")["defending"].mean().sort_values(ascending=False).head(5).index.tolist())
```

## <span style="color:red"> **Question 3** </span>

##### The last question we want to answer: 
##### How do offensive strategies (shooting) and defensive strategies (defending) correlate for men in 2022 by country?

- ##### For this question, we will look at the players_22 dataset containing men's player-level performance attributes from the FIFA 22 video game.

### <span style="color:red"> **Description** </span>

##### We will be reading in the fifa data set, players_2022. The dataset contains 19,239 observations, where each row represents a single male soccer player included in the FIFA 22 video game. Each row provides detailed information about a player, including nationality, club affiliation, physical characteristics, and more than 100 performance-related statistics such as shooting, passing, dribbling, and defending. For the purpose of this research question, I focus specifically measuring shooting and defending ability, as these attributes allow us to examine how offensive and defensive strengths vary across countries. I begin by importing the necessary libraries, loading the dataset, and confirming the number of observations.

```python
# Read in our data set
players = pd.read_csv("2-FIFA/players_22.csv")

# Count total observations
players.shape[0]
```

### <span style="color:red"> **Cleaning and Description** </span>

##### Before conducting the analysis, I perform several data cleaning steps to ensure the dataset is in a usable format. First, I select only the columns relevant to my research question: player name, nationality, shooting rating, and defending rating. I then remove any rows with missing shooting or defending values to avoid distortions in the country-level averages. Finally, I rename the columns to shorter, clearer labels that simplify the analysis process. These steps ensure that the dataset contains only complete observations and variables directly tied to evaluating offensive and defensive performance across nations.

```python
# Select only the columns relevant to this analysis
df = players[['short_name','nationality_name','shooting','defending']]

# Remove rows with missing shooting or defending values
df = df.dropna(subset=['shooting','defending'])

# Rename columns to shorter, clearer names
df = df.rename(columns={
    'nationality_name': 'country',
    'shooting': 'shoot',
    'defending': 'defend'
})

# Display cleaned data
df.head()
```

##### The main variables used in this analysis are shooting and defending, which represent a player’s offensive and defensive strengths respectively. Shooting combines factors such as finishing, shot power, and accuracy, while defending includes attributes like tackling, marking, and defensive awareness. These two attributes provide a clear and focused way to evaluate offensive and defensive capabilities across countries. To better understand how these values are distributed across all players, I compute summary statistics—including the minimum, maximum, mean, and standard deviation—which help contextualize each variable prior to aggregating the scores at the country level.

```python
# Summary statistics for main variables
df[['shoot', 'defend']].describe()
```

##### To standardize the shooting and defending attributes, I apply min–max normalization, which rescales each value between 0 and 1 based on that column’s observed minimum and maximum. Under this method, the lowest value becomes 0, the highest becomes 1, and all other values fall proportionally between those extremes. This approach enables direct comparison of the two attributes on the same scale without altering their underlying distribution. This transformation maintains relative ordering while making the variables easier to work with in later visualizations.

```python
# Define a function to scale numeric values to a 0–1 range
def scale(x):
    return (x - x.min()) / (x.max() - x.min())

# Apply scaling transformation to create normalized shooting and defending columns
df['shoot_scaled'] = scale(df['shoot'])
df['defend_scaled'] = scale(df['defend'])

# Preview the dataset after transformation
df.head()
```

##### To answer the research question, I compute the average shooting and defending values for each country in the dataset. By grouping players by their nationality and calculating the mean shooting and defending ratings, we obtain a country-level perspective on offensive and defensive strength. I then sort the results by average shooting rating to make it easier to compare countries with stronger or weaker offensive profiles. These grouped averages form the basis for the scatter plot and correlation analysis that follow.

```python
# Group data by country and compute mean shooting and defending
country_stats = (
    df.groupby("country")[["shoot","defend"]]
      .mean()  # compute averages
      .reset_index()  # convert index back to column
)

# Display the top 10 countries by average shooting
country_stats.head(10)
```

##### Next, I define a custom metric called the balance score, which measures how similar a country’s average shooting and defending values are. The score ranges from 0 to 1, where values closer to 1 indicate that a country’s offensive and defensive abilities are more evenly matched. After defining the function, I compute the score for each country using a loop that iterates through the rows of the aggregated dataset.

```python
# Define a function that measures how similar shooting and defending are
def balance_score(row):
    return 1 - abs(row['shoot'] - row['defend']) / 100

# Create an empty list to store balance scores
scores = []

# Loop through each row of the grouped dataset and compute the score
for _, row in country_stats.iterrows():
    scores.append(balance_score(row))

# Add the balance score as a new column
country_stats['balance_score'] = scores

# View results
country_stats.head()
```

### <span style="color:red"> **Graphs and Findings** </span>

##### To visualize the relationship between average offensive and defensive strengths across countries, I create a scatter plot with average defending on the x-axis and average shooting on the y-axis. Each point represents a country. I also compute a line of best fit to show the overall trend in the data. This regression line helps indicate whether the relationship is positive, negative, or negligible.

```python
# Extract x and y for convenience
x = country_stats['defend']
y = country_stats['shoot']

# Scatter plot
plt.figure(figsize=(12,7))
plt.scatter(x, y, alpha=0.6)

# Compute regression line (line of best fit)
slope, intercept = np.polyfit(x, y, 1)
line = slope * x + intercept

# Plot the regression line
plt.plot(x, line, color='red', label=f"Best fit line (slope = {slope:.2f})")

# Add title and axis labels
plt.title("Relationship Between Offensive (Shooting) and Defensive Strength by Country (FIFA 22)")
plt.xlabel("Average Defending (Country)")
plt.ylabel("Average Shooting (Country)")
plt.legend()

# Add grid lines for readability
plt.grid(alpha=0.3)

plt.show()
```

##### The scatter plot shows substantial variation in how countries perform offensively and defensively. The regression line slopes slightly downward, indicating a weak negative relationship between average shooting and defending ratings. This suggests that countries with higher defending values are not necessarily the same countries with higher shooting values. Instead, countries tend to differ in whether their strengths lean more toward offensive or defensive attributes. While the correlation is not strong, the trend line helps highlight an overall pattern where increases in one ability do not consistently align with increases in the other.

##### Overall, these results suggest that national playing styles and development priorities may influence the types of players a country produces. Rather than developing uniformly strong all-around players, countries may emphasize specific attributes that align with their coaching styles, or tactical approaches. For example, some nations may concentrate on developing technically skilled attackers, while others may focus more on strong defensive fundamentals.
